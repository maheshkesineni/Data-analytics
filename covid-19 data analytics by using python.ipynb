```python
x=df.iloc[61:,0].values
x
```

```python
y=df.iloc[61:,1].values
y
```

```python
plt.figure(figsize=(25,8))
ax=plt.axes()
ax.set_facecolor('black')
ax.set_xlabel('\n date',size=25)
ax.set_ylabel('\n number of cases',size=25)
for i,j in zip(x,y):
    ax.annotate(j,xy=(i,j),color='white',size=15)
ax.annotate('second lockdown 15 appear 2021',
            xy=(15.5,860),
            xytext=(19.9,500),
            size=15,
            color='white',
            arrowprops=dict(color='white',linewidth=1.0)
           )
    

plt.xticks(rotation='vertical',size=15)
plt.yticks(size=15)
plt.title('covid-19 data analytics by using python',size=30)
plt.plot(x,y,marker='o',ms=10)
```
