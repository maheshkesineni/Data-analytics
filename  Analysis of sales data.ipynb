```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Sales Data.csv',encoding='unicode_escape')
df
```

```python
df.shape
```

```python
#drop: we can delete the coloumns
df.drop(['Status',	'unnamed1'],axis=1,inplace=True)
df
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
#dropna(): remove null values
df.dropna(inplace=True)
df
```

```python
#discount
df['Discount']=df['Amount']*0.1
df
```

```python
#summary of statistics
df.describe()
```

```python
#rename():is funtion which is used to rename of the given column
#df.rename(columns={'old':'new'})
df.rename(columns={'Marital_Status':'Married'},inplace=True)
df
```

```python
top_10=df.head(10)
top_10
```

```python
#plot graph based on gender wise sales(amount)

plt.figure(figsize=(12,5))
plt.xlabel("Gender")
plt.ylabel('Amount')
plt.plot(top_10['Gender'],top_10['Amount'],marker='o')
plt.grid()

```

```python
#plot graph based on age wise sales(amount)

plt.figure(figsize=(12,5))
plt.xlabel("Age")
plt.ylabel('Amount')
plt.bar(top_10['Age'],top_10['Amount'])

```

```python
#plot graph based on age group wise sales(amount)

plt.figure(figsize=(12,5))
plt.xlabel("Age group")
plt.ylabel('Amount')
plt.bar(top_10['Age Group'],top_10['Amount'])
```

```python
#plot graph based on state wise sales(amount)

plt.figure(figsize=(12,5))
plt.xlabel("atate")
plt.ylabel('Amount')
plt.bar(top_10['State'],top_10['Amount'])
```

```python
#plot graph based on zone wise sales(amount)

plt.figure(figsize=(12,5))
plt.xlabel("zone")
plt.ylabel('Amount')
plt.plot(top_10['Zone'],top_10['Amount'])
```

```python
fig,ax=plt.subplots(nrows=1,ncols=1,figsize=(12,5))
barwidth=0.25
bar1=[i+barwidth for i in range(len(top_10['Amount']))]
bar2=[i+barwidth for i in bar1]
plt.bar(bar1,top_10['Amount'],width=barwidth,label='Amount')
plt.bar(bar2,top_10['Discount'],width=barwidth,label='Discount')
plt.xticks([i+barwidth for i in range(len(top_10['State']))],
          (top_10['State']),rotation='vertical')
plt.legend()
plt.show()
```

```python
df['Final_amount']=df['Amount']-df['Discount']
```

```python
df
```

```python
top_final=df.head(10)
top_final
```

```python
fig,ax=plt.subplots(nrows=1,ncols=1,figsize=(12,5))
barwidth=0.25
bar1=[i+barwidth for i in range(len(top_10['Amount']))]
bar2=[i+barwidth for i in bar1]
plt.bar(bar1,top_final['Amount'],width=barwidth,label='Amount')
plt.bar(bar2,top_final['Final_amount'],width=barwidth,label='Final_amount')
plt.xticks([i+barwidth for i in range(len(top_10['State']))],
          (top_10['State']),rotation='vertical')
plt.legend()
plt.show()
```

```python
plt.pie(top_final['Amount'],labels=top_final['Age Group'])
plt.show()
```

```python
fig,ax=plt.subplots(2,2,figsize=(20,10))
ax[0,0].plot(top_10['Gender'],top_10['Amount'],marker='o')
ax[0,0].set_title("Gender-wise Sales Amount")
ax[0,0].set_xlabel("Gender")
ax[0,0].set_ylabel("Amount")


ax[0,1].bar(top_10['Age'],top_10['Amount'])
ax[0,1].set_title("Age-wise Sales Amount")
ax[0,1].set_xlabel("Age")
ax[0,1].set_ylabel("Amount")


ax[1,0].bar(top_10['Age Group'],top_10['Amount'])
ax[1,0].set_title("Age Group-wise Sales Amount")
ax[1,0].set_xlabel("Age Group")
ax[1,0].set_ylabel("Amount")


 

barwidth = 0.25
bar1 = [i + barwidth for i in range(len(top_final['Amount']))]
bar2 = [i + barwidth for i in bar1]

ax[1,1].bar(bar1, top_final['Amount'], width=barwidth, label='Amount')
ax[1,1].bar(bar2, top_final['Final_amount'], width=barwidth, label='Final_amount')

plt.xticks([i+barwidth for i in range(len(top_10['Zone']))],
          (top_10['Zone']),rotation='vertical')

ax[1,1].legend()

plt.show()




```
