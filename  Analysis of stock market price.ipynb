```python
top10=df.head(10)
top10
```

```python
fig,ax=plt.subplots(2,2,figsize=(20,10))
ax[0,0].plot(top10['date'],top10['close'])
ax[0,0].tick_params(axis='x', rotation=90)

bars=ax[0,1].bar(top10['date'],top10['close'])
ax[0,1].tick_params(axis='x', rotation=90)
for bar in bars:
    yval=bar.get_height()
    ax[0,1].text(bar.get_x()+bar.get_width()/2.0,yval+6,yval,ha='center')
ax[0,1].set_ylim(0,max(top10['open'])+100)

ax[1,0].pie(top10['high'],labels=top10['date'])
plt.show()
plt.tight_layout()
```
