```python
# analysis of monthly sales revenue
import matplotlib.pyplot as plt
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
revenue = [45, 52, 49, 60, 65, 58, 72, 78, 70, 85, 92, 110] # In thousands
colors=['red', 'green', 'blue', 'orange', 'purple', 'cyan', 'magenta', 'yellow', 'brown', 'pink', 'lime', 'teal']
plt.plot(months,revenue,marker="o",ms=3,color="red")
plt.title("Month wise revenue")
plt.xlabel("months")
plt.ylabel("revenue")
plt.savefig("monthly sales revenue.jpg")
plt.grid()
plt.show()
```

```python
# analysis of categories wise sales in 2025
categories = ['Electronics', 'Apparel', 'Home & Kitchen', 'Beauty', 'Sports']
sales_vols = [320, 240, 180, 110, 95]
plt.plot(categories,sales_vols,marker="o",ms=3,color="red")
plt.title("categories wise sales in 2025")
plt.xlabel("categories")
plt.ylabel("sales_vols")
plt.savefig("categories wise sales in 2025.jpg")
plt.grid()
plt.show()
```

```python
# month wise sales for 3(2023,2024,2025) years -- plot  --- 2024 2025 2026 -- legend()
Month=['jan','feb','march','apr','may','june']
y1= [100, 120, 110, 130, 150, 140]
y2= [110, 130, 125, 145, 160, 155]
y3= [130, 150, 140, 170, 190, 180]
plt.plot(Month,y1,label=2023)
plt.plot(Month,y2,label=2024)
plt.plot(Month,y3,label=2025)
plt.legend()
plt.title("Month wise sales")
plt.xlabel("Month")
plt.ylabel("sales_vols")
plt.savefig("month wise sales.jpg")
plt.grid()
plt.show()
```

```python
#  Sales Representative Performance 
# ----------------------------------------------------
product=['lenovo','dell','hp','sony']
actual_sales = [95, 110, 85, 120]
target_sales = [100, 100, 90, 110]
plt.plot(product,actual_sales,label='actual_sales')
plt.plot(product,target_sales,label='target_sales')
 
plt.legend()
plt.title("Sales Representative Performance")
plt.xlabel("product")
plt.ylabel("sales_vols")
plt.savefig("Sales Representative Performanc.jpg")
plt.grid()
plt.show()
```

```python
#Analysis of covid-19 country wise Cases,Recovery and Death
country=['India','China','USA','Japan','Australia','UK']
Cases=[12000,3000,20000,50000,1000,8000]
Recovery=[1300,1200,900,200,1000,600]
Deth=[1000,1200,400,6000,1800,2100]
plt.plot(country,Cases,label="Cases")
plt.plot(country,Recovery,label="Recovery")
plt.plot(country,Deth,label="Deth")
plt.title("Analysis of covid-19 country wise Cases,Recovery and Death")
plt.xlabel("country")
plt.ylabel("Cases,Recovery and Death")
plt.legend()
plt.grid()
plt.show()
plt.savefig("Analysis of covid-19 country wise Cases,Recovery and Death.jpg")
```
